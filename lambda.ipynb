```python
#1.sorting-Ascending

numbers = [45,12,78,3,29,56,8]
numbers.sort(key=lambda x:x)
numbers
```

```python
#2.sort by string length shortest to longest

words = ["python","ai","machine","data","learning","ml"]
words.sort(key=lambda x:len(x))
words
```

```python
#3.sort-descending

numbers = [15,42,7,89,23,64,31]
numbers.sort(key=lambda x:x,reverse=True)
numbers
```

```python
#4.filtering even numbers

numbers = [12,7,25,40,33,18,9,50]
even=list(filter(lambda x:x %2 ==0,numbers))
even
```

```python
#5.filtering-numbers greater than 50.

numbers=[25,67,12,89,45,73,34,91]
num=list(filter(lambda x:x>50,numbers))
num
```

```python
#6.mapping-square numbers

numbers=[2,4,6,8,10]
squares=list(map(lambda x:x**2,numbers))
squares
```

```python
#7.mapping-convert to positive

numbers=[-10,5,-7,12,-3,8]
num=list(map(lambda x:abs(x),numbers))
num
```

```python
#8.reducing-sum find sum

from functools import reduce
numbers = [10,20,30,40,50]
sum=reduce(lambda x,y: x+y,numbers)
sum
```

```python
#9.reducing-maximum number find largest numm

from functools import reduce
numbers = [45,12,89,34,67,23]
largest=reduce(lambda a,b: numbers)
largest
```
