```python
#1.Create an array 1 to and print it 

import numpy as np

arr=np.arange(1,11)
arr
```

```python
#2.Array operations

a=np.array([10,20,30,40,50])
b=np.array([1,2,3,4,5])

#Addition
print("Addition")
print(a+b)

#Substraction
print("Substraction")
print(a-b)

#Multiplication
print("Multiplicaton")
print(a*b)

#Division
print("Division")
print(a/b)
```

```python
#3.find maximum and minimum

marks=np.array([85,72,91,65,78,88])

print("Maximum Mark:", np.max(marks))
print("Minimum Mark", np.min(marks))
print("Average mark:", np.average(marks))
```
