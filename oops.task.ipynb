```python
#2.class method car with attribute brand and model

class Car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
car1=Car("Toyota", "Innova")
print("brand", car1.brand)
print("model",car1.model)
```

```python
#3.constructor init 

class Employee:
    def __init__(self,name, department, salary):
        self.name= name
        self.department = department
        self.salary = salary

    def Display(self):
        print("Name:", self.name)
        print("departmet:",self.department)
        print("salary:",self.salary)

employee1=Employee("Anu", "IT", 30000)
employee1.Display()
```

```python
#4.bank account

class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
    def func(a):
        print("Account Holder:", a.account_holder)
        print("Balance:", a.balance)
b1=BankAccount("Arun", 25000)
b1.func()
b2=BankAccount("Rahul", 40000)
b2.func()
b3=BankAccount("Neha", 35000)
b3.func()
```

```python
#5.rectangle

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(a):
     result=a.length* a.width
     return result
a1=Rectangle(10,5)
print("Area:", a1.area())

    
```

```python
#6. division by 0


try:
    a=int(input('Enter numerator:'))
    b=int(input('Enter denominator:'))
    print(a/b)
except ZeroDivisionError:
    print('Cannot divided by zeroo')
except NameError:
    print('exception occured in nameerrror')
except ValueError:
    print('Exception occured in valueerror')
except:
    print('an exception occured')
else:
    print('x printed')
finally:
    print('successfully completed')
```

```python
#7. handle invalid input

try:
    a=int(input('Enter numerator:'))
    b=int(input('Enter denominator:'))
    print(a/b)
except ZeroDivisionError:
    print('Cannot divided by zeroo')
except NameError:
    print('exception occured in nameerrror')
except ValueError:
    print('invalid input!! Please enter a number')
except:
    print('an exception occured')
else:
    print('x printed')
finally:
    print('successfully completed')
```

```python
#8. multiple exceptions 

try:
    a=int(input('Enter numerator:'))
    b=int(input('Enter denominator:'))
    print(a/b)
except ZeroDivisionError:
    print('Cannot divided by zeroo')
except ValueError:
    print('invalid input!! Please enter a number')
```

```python
#9. list index exception

numbers=[10,20,30,40,50]
try:
    index=int(input('Enter the index:'))
    print('Element at this index is',numbers[index])
except IndexError:
    print('Invalid index')
except ValueError:
    print('ivalid input!')
```
