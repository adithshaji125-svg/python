```python
#logical

1.
marks={"Anu":85,"Rahul":72,"Akhil":91,"Meera":65,"Neha":78}
highest_student=max(marks, key=marks.get)
highest_mark=marks[highest_student]
print(f"1. Student with highest mark: {highest_student} - {highest_mark}")
2.
lowest_student=min(marks, key=marks.get)
lowest_mark=marks[lowest_student]
print(f"2. Student with lowest mark: {lowest_student} - {lowest_mark}")
3.
average_mark=sum(marks.values()) / len(marks)
print(f"3. Average mark: {average_mark}")
4.
above_75={student: mark for student, mark in marks.items() if mark >= 75}
print(f"4. Students with 75 or above: {above_75}")

```

```python
#2 pattern printing


for i in range(1,5):
    for j in range(1,i+1):
      print(j,end="")
    print()

for i in range(5,0,-1):
    for j in range(1,i+1):
        print(j,end="")
    print()



```

```python
#machine test
#ATM withdrawal

balance=10000
withdrawal=int(input("Enter withdrawal amount"))

if withdrawal > 0:
    if withdrawal <= balance:
        remaining_balance = balance - withdrawal
        print("Withdrawal successful")
        print(f"Remaining balance: ${remaining_balance}")
    else:
        print("Insufficient balance. Please enter a valid amount")
else:
    print("Withdrawal amount must be greater than zero")

```
