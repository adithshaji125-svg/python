```python
#match statements

day=10
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("no found")
```

```python
a = int(input("enter value a"))
b = int(input("enter value b"))

op = input("enter operator(+-*)")

match op:
    case "+":
        print(a+b)
    case "-":
        print(a-b)
```

```python
day=6
month=2

match day:
    case 1  | 2|3|4|5 if month ==1:
        print("weekday in jan")

    case 6|7 if month==1:
        print("weekend in jan")

    case 1 | 2|3|4|5 if month ==2:
        print("weekday in feb")

    case 6|7 if month==2:
        print("weekend in feb")
```
