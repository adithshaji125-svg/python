```python
#1.check a num odd or even using shorthand if

number = int(input("enter a number"))
print("Even") if number % 2 == 0 else print("Odd")
```

```python
#6.write a program to display day 1 to 7 (num entered by the user)

day=int(input("Enter the day"))
match day:
  case 1:
        print("Monday")
  case 2:
        print("Tuesday")
  case 3:
      print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
```

```python
#7.create calculator perform(addition,substraction,multiplication or division)

a=int(input('Enter first number'))
b=int(input("Enter second number"))
op = input("enter operator(+-*/)")

match op:
    case "+":
        print(a+b)
    case "-":
        print(a-b)
    case "*":
        print(a*b)
    case "/":
        print(a/b)
```

```python
#8. Grade description(A,B,C,D,F)

grade=input("Enter your grade A, B, C, D, F ")

match grade:
    case "A":
        print("Excellent")
    case "B":
        print("Good")
    case "C":
        print("Average")
    case "D":
        print("Poor")
    case "F":
        print("Fail")
    case _:
        print("no found")
```

```python
#9.display action for traffic light(Red,yellow,green)

signal=input("Enter signal (Red, Yellow, Green) ")

match signal:
    case "red":
        print("Stop")
    case "yellow":
        print("Go slowly")
    case "green":
        print("Go")
    case _:
        print("invalid")
```
